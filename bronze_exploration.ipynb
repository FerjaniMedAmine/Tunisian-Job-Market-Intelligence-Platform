```python
import json
from pathlib import Path

import pandas as pd
from bs4 import BeautifulSoup
```

```python
#saving bronze paths in a variable "records"



REPORT_PATH = Path(
    "data/reports/keejob/bfce5dcc9bf046eaaa8aff9070e0a269.json"
)

with REPORT_PATH.open(encoding="utf-8") as file:
    report = json.load(file)

records = []
df = pd.json_normalize(records, sep=".")

print("Rows and columns:", df.shape)
df.head()
for saved in report["saved_records"]:
    with Path(saved["bronze_path"]).open(encoding="utf-8") as file:
        records.append(json.load(file))

print(f"Loaded {len(records)} observations")
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
```

```python
df = pd.json_normalize(records, sep=".")

print("Rows and columns:", df.shape)
df.head()
```

```python
df.drop(columns=["payload.@type","payload.identifier.@type","payload.hiringOrganization.@type","payload.jobLocation.@type","payload.jobLocation.address.@type","payload.baseSalary.@type","payload.baseSalary.value.@type","payload.applicantLocationRequirements.@type"],inplace=True)

df.drop(columns=["payload.hiringOrganization.logo"],inplace=True)
df.drop(columns=["payload.applicantLocationRequirements.name"],inplace=True)

df.drop(columns=["metadata.requested_url","metadata.response_url","metadata.http_status","metadata.content_type","payload.@context"],inplace=True)
```

```python
df.drop(df[df["payload.jobLocation.address.addressCountry"]!="TN"].index,inplace=True)
```

```python
date_formats = {
    "payload.datePosted": "%Y-%m-%d",
    "payload.validThrough": "%Y-%m-%dT%H:%M:%S",
}

for column, date_format in date_formats.items():
    df[column] = pd.to_datetime(
        df[column],
        format=date_format,
        errors="raise",
    )

df["metadata.collected_at"] = pd.to_datetime(
    df["metadata.collected_at"],
    utc=True,
    errors="raise",
)

invalid_dates = df.loc[
    df["payload.validThrough"] < df["payload.datePosted"],
    [
        "payload.identifier.value",
        "payload.title",
        "payload.datePosted",
        "payload.validThrough",
    ],
]

print("Expiry earlier than publication:", len(invalid_dates))
display(invalid_dates)
```

```python
from bs4 import BeautifulSoup


def extract_description_text(html):
    if pd.isna(html):
        return pd.NA

    soup = BeautifulSoup(html, "html.parser")

    for tag in soup(["script", "style"]):
        tag.decompose()

    text = soup.get_text(separator=" ", strip=True)

    return " ".join(text.split())


df["payload.description"] = df["payload.description"].map(
    extract_description_text
)

```
